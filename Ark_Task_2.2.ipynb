```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load images
ironman = cv2.imread("iron_man_noisy.jpg")
scenery = cv2.imread("noisy.jpg")

# Convert to grayscale
ironman_gray = cv2.cvtColor(ironman, cv2.COLOR_BGR2GRAY)
scenery_gray = cv2.cvtColor(scenery, cv2.COLOR_BGR2GRAY)

plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.title("Iron Man Noisy")
plt.imshow(ironman_gray, cmap='gray')
plt.axis("off")

plt.subplot(1,2,2)
plt.title("Scenery Noisy")
plt.imshow(scenery_gray, cmap='gray')
plt.axis("off")

plt.show()
```

```python
ironman_median = cv2.medianBlur(ironman_gray, 5)  
scenery_median = cv2.medianBlur(scenery_gray, 5)
```

```python
kernel = np.ones((3,3), np.uint8)

ironman_clean = cv2.morphologyEx(ironman_median, cv2.MORPH_OPEN, kernel)
scenery_clean = cv2.morphologyEx(scenery_median, cv2.MORPH_OPEN, kernel)
```

```python
ironman_edges = cv2.Canny(ironman_clean, 50, 150)
scenery_edges = cv2.Canny(scenery_clean, 50, 150)
```

```python
plt.figure(figsize=(12,6))

plt.subplot(2,2,1)
plt.title("Iron Man Cleaned")
plt.imshow(ironman_clean, cmap='gray')
plt.axis("off")

plt.subplot(2,2,2)
plt.title("Scenery Cleaned")
plt.imshow(scenery_clean, cmap='gray')
plt.axis("off")

plt.subplot(2,2,3)
plt.title("Iron Man Edges")
plt.imshow(ironman_edges, cmap='gray')
plt.axis("off")

plt.subplot(2,2,4)
plt.title("Scenery Edges")
plt.imshow(scenery_edges, cmap='gray')
plt.axis("off")

plt.show()
```

```python
cv2.imwrite("ironman_denoised.jpg", ironman_clean)
cv2.imwrite("scenery_denoised.jpg", scenery_clean)

cv2.imwrite("ironman_edges.jpg", ironman_edges)
cv2.imwrite("scenery_edges.jpg", scenery_edges)
```
